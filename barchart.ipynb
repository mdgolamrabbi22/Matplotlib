```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
```

```python
category = np.array(["Electronic", "Grosaries", "Book", "Pen", "Laptop", "Battery"])
quantity = np.array([2,5,8,4,0,1])
```

```python
# Bar Charts 
plt.bar(category,quantity, color="green" )
# plt.barh(category,quantity, color="green" )
plt.title("Product Sell Data",color="#18253d",font="arial",fontsize=20)
plt.xlabel("Product Catagory", color="#1f5bcc",font="arial",fontsize=20)
plt.ylabel("Quantity", color="#1f5bcc",font="arial",fontsize=20)
plt.show()
```

```python
# Pie charts 
category = ["Fresser", "First", "Secound", "Third", "Forth"]
values = np.array([300,150,80,250,180])
color = ["green", "red", "blue", "gray", "yellow"]
plt.pie(values ,
        labels=category,
        colors=color,
        autopct="%1.1f%%",
        explode=[0,0,0,0.1,0],
        shadow=True,
        startangle=90)
plt.title("Collage Student Data", color="#0ff5b4",font="arial",fontsize=20)
plt.show()
```

```python
# Scatter chart
# ১. পজিটিভ কোরিলেশন (পড়াশোনা বাড়লে গ্রেড বাড়ে)
study_time_1 = np.array([1, 2, 3, 5, 6, 8, 9])
grades_1 = np.array([30, 45, 50, 65, 75, 85, 95])

# ২. রিয়েলিস্টিক ডাটা (বাস্তবসম্মত ছড়ানো ডাটা)
study_time_2 = np.array([1, 2, 4, 4, 6, 7, 9])
grades_2 = np.array([40, 35, 65, 50, 60, 85, 75])

# ৩. নো কোরিলেশন (কোনো সম্পর্ক নেই)
study_time_3 = np.array([2, 3, 5, 5, 7, 8, 9])
grades_3 = np.array([80, 40, 85, 35, 90, 45, 80])
```

```python
plt.scatter(study_time_1,grades_1,color="red",alpha=0.8, s=150,marker="s" ,label="ClassA")
plt.scatter(study_time_2,grades_2,color="green",alpha=0.8, s=150,marker="o", label="ClassB")
plt.scatter(study_time_3,grades_3,color="blue",alpha=0.8, s=150,marker="^", label="ClassC")
plt.title("Study Hours and Grades data",color="#0ff5b4",font="arial",fontsize=20)
plt.xlabel("study Hours",color="#1f5bcc",font="arial",fontsize=20)
plt.ylabel("Grades",color="#1f5bcc",font="arial",fontsize=20)
plt.legend()
plt.show()
```

```python
# Histragram chart
scores = np.random.normal(loc=80,scale=10,size=100)
scores = np.clip(scores, 0, 100)

plt.hist(scores,
         bins=10,
         color="green",
         edgecolor="black"
         )
plt.title("Student Exam Scores")
plt.xlabel("Scores")
plt.ylabel("# of Student")
plt.show()
```
